# 🎬 Zero-Cost AI Video Studio (Colab T4 Launcher)
You can run this notebook either:
1. **Directly inside your IDE (Local -> Colab Remote Kernel)** using the **Google Colab extension**, OR
2. **In the Colab Web UI**.

It safely switches to `/content` first, clones [`menikhilravi/ai-video-gen`](https://github.com/menikhilravi/ai-video-gen.git), removes Colab's conflicting `torchao` package, installs dependencies cleanly, and launches the Gradio `--share` URL.

```
# @title 🚀 1-Click GitHub Clone & Launch
import os
import shutil
import torch

# Always move to /content first so deleting an old repo folder never invalidates cwd
os.makedirs("/content", exist_ok=True)
os.chdir("/content")

print("🔍 Checking GPU availability...")
if torch.cuda.is_available():
    gpu_name = torch.cuda.get_device_name(0)
    vram_gb = torch.cuda.get_device_properties(0).total_memory / (1024 ** 3)
    print(f"✅ GPU Detected: {gpu_name} ({vram_gb:.1f} GB VRAM) | PyTorch {torch.__version__}")
else:
    print("⚠️ No CUDA GPU detected! Make sure your Colab runtime is set to T4 GPU.")

# 1. Fresh clone of your GitHub repository
REPO_DIR = "/content/ai-video-gen"
if os.path.exists(REPO_DIR):
    shutil.rmtree(REPO_DIR)

!git clone https://github.com/menikhilravi/ai-video-gen.git /content/ai-video-gen
os.chdir(REPO_DIR)

# 2. Remove Colab's conflicting torchao package & install requirements
print("📦 Cleaning conflicting torchao & installing dependencies...")
!pip uninstall -y -q torchao
!pip install -q -r /content/ai-video-gen/requirements.txt

# 3. Launch Gradio with public share link
!python3 /content/ai-video-gen/app.py --share
```
